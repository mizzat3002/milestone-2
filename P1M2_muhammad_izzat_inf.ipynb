# Model Inference

### Import Libraries

```python
import joblib
import pandas as pd
```

### Load Model

```python
model = joblib.load('model_stroke.pkl')
```

### Data pasien baru

```python
data_pasien_baru = {
    'gender': ['Male'],
    'age': [67.0],
    'hypertension': [1],       
    'heart_disease': [1],      
    'ever_married': ['Yes'],
    'work_type': ['Private'],
    'residence_type': ['Rural'],
    'avg_glucose_level': [228.69],
    'body_mass': [36.6],
    'smoking_status': ['formerly smoked']
}

df_inf = pd.DataFrame(data_pasien_baru)
df_inf
```

### Prediction

```python
prediction = model.predict(df_inf)

# menampilkan hasil prediksi
df_inf['prediction'] = prediction
df_inf['status'] = df_inf['prediction'].map({1: 'Resiko Stroke', 0: 'Sehat'})
print(df_inf[['gender', 'age', 'status']])
```
